### 미션 2

```python
#코랩용 : 그래프의 선 모양, 색깔, 마커 모양은 아래 코드와 달라도 됨

import matplotlib.pyplot as plt
import pandas as pd

plt.rc('font', family='NanumGothic')
df=pd.read_excel('data1.xlsx')

plt.plot(df['name'], df['kor'], 'k--', marker='o', label='국어점수')
plt.plot(df['name'], df['math'], marker='v',label='수학점수')
plt.title('성적 그래프')
plt.xlabel('이름')
plt.ylabel('점수')
plt.legend()
plt.show()
```

```python
# 주피터노트북용 :그래프의 선 모양, 색깔, 마커 모양은 아래 코드와 달라도 됨

import matplotlib.pyplot as plt
import pandas as pd

plt.rc('font', family='Gulim')
df=pd.read_excel('data1.xlsx')

plt.plot(df['name'], df['kor'], 'k--', marker='o', label='국어점수')
plt.plot(df['name'], df['math'], marker='v',label='수학점수')
plt.title('성적 그래프')
plt.xlabel('이름')
plt.ylabel('점수')
plt.legend()
plt.show()
```

### 미션3

```python
# 미션 3-1)
import matplotlib.pyplot as plt
import seaborn as sns

penguins_data = sns.load_dataset('penguins')
sns.barplot(x = 'species', y = 'body_mass_g', data=penguins_data, hue='sex')
plt.show()
```

```python
# 미션 3-2)
sns.pairplot(data=penguins_data)
plt.show()
```

```python
# 미션 3-3)
sns.regplot(data=penguins_data, x='body_mass_g', y='flipper_length_mm')
plt.show()
```
