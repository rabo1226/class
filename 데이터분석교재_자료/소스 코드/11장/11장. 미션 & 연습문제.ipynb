## 미션

```python
#미션
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

df = pd.read_excel('사과배복숭아-학습.xlsx')
df.fillna(0, inplace=True)
df.replace({'fruit':{ '사과':0, '배':1, '복숭아':2}}, inplace=True)
x_train = df.iloc[:, 1:]  
y_train = df.iloc[:,0]    

model_3 = RandomForestClassifier(n_estimators=3,random_state=0)
model_3.fit(x_train, y_train)

model_9 = RandomForestClassifier(n_estimators=9,random_state=0)
model_9.fit(x_train, y_train)

model_15 = RandomForestClassifier(n_estimators=15,random_state=0)
model_15.fit(x_train, y_train)

df_test = pd.read_excel('사과배복숭아-테스트.xlsx')
df_test.fillna(0, inplace=True)
df_test.replace({'fruit':{'사과':0, '배':1, '복숭아':2}}, inplace=True)
x_test =  df_test.iloc[:, 1:]   
y_test =  df_test.iloc[:,0]     

prd_3 = model_3.predict(x_test)
prd_9 = model_9.predict(x_test)
prd_15 = model_15.predict(x_test)

# 예측값과 테스트 데이터의 정답을 비교해서 정확도 계산
print('정확도(트리 3개) =',accuracy_score(y_test, prd_3))
print('정확도(트리 9개) =',accuracy_score(y_test, prd_9))
print('정확도(트리 15개) =',accuracy_score(y_test, prd_15))
```

### 6번

```python
import pandas as pd
import numpy as np
df = pd.read_excel('사과배복숭아-학습.xlsx')
df.fillna(0, inplace=True)
df.replace({'fruit':{'사과':0, '배':1, '복숭아':2}}, inplace=True)
x_train = df.iloc[:, 1:]  # 학습 데이터의 속성값
y_train = df.iloc[:,0]    # 학습 데이터의 정답
df_test = pd.read_excel('사과배복숭아-테스트.xlsx')
df_test.fillna(0, inplace=True)
df_test.replace({'fruit':{'사과':0, '배':1, '복숭아':2}}, inplace=True)
x_test =  df_test.iloc[:, 1:]   
y_test =  df_test.iloc[:,0]     

from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier(random_state=0)
model.fit(x_train, y_train)

# 연습문제6
from sklearn.metrics import accuracy_score
prd = model.predict(x_test)
print('정확도 =',accuracy_score(y_test, prd))
```
