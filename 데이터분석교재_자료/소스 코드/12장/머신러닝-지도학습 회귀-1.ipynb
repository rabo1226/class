# 1.선형 회귀(Linear Regression) 알고리즘 

### 데이터 전처리

```python
#데이터 저장
import numpy as np

# 학습 데이터(재배 면적)로 전달되는 독립변수(조건 칼럼)은 2차원 배열이어야 함
x = np.array( [[2.1], [10], [3], [1], 
               [3.5], [5], [8]] , dtype=float)
# 정답 데이터 (수확량)
y = np.array( [64.9, 292.6, 85.9, 30.92, 
               110.5, 163.4, 230.1] , dtype=float)

print("재배 면적")
print(x)
print("수확량 :",y)
```

```python
# 그래프로 살펴보기
import matplotlib.pyplot as plt

plt.scatter(x, y)
plt.show()
```

### 모델 생성 및 학습

```python
#머신 러닝 학습
from sklearn.linear_model import LinearRegression

#학습 모델 객체 생성
LR_model = LinearRegression()

#학습 데이터를 전달하여 학습 모델 완성
LR_model.fit(x, y)
```

```python
#생성한 모델을 사용해서 예측
#재배 면적 7과 12 를 전달했을 때 예상되는 귤 수확량 계산
prd = LR_model.predict([[7], [12]])
prd
```

```python
#그래프로 살펴보기
# 학습 데이터 표시
plt.scatter(x, y, label='Train Data')

# 예측한 데이터 표시
plt.scatter([7, 12], prd, marker='v', label='Predicted Data')
plt.legend()
plt.show()
```

```python
#그래프에 추세선 추가
#기울기 : coef_ , y절편 : intercept_

line_x = np.array([1, 12]) 
line_y = line_x*LR_model.coef_+LR_model.intercept_
plt.plot(line_x,line_y)

plt.scatter(x, y, label='Train Data')
plt.scatter([7, 12], prd, marker='v', label='Predicted Data')
plt.legend()
plt.show()
```
