```python
### 서식 문자열
```

```python
a="파이썬"
b=20
print("%s아 반가워~"%a)
print("%s아 반가워~ %d살 이구나~"%(a,b))
```

```python
a=24; b=21; c=(a+b)/2

print(c)
print("%d"%c)
print("%4d"%c)
print()
print("%f"%c)
print("%3.1f"%c)
print("%6.2f"%c)
```

```python
print("오늘은 {}월 {}일 입니다.".format(6,26))
```

```python
print("오늘은 {:d}월 {:5.2f}일 입니다.".format(6,26))
```

```python
a=['파이썬','머신러닝','딥러닝']
print('{} {} {}'.format(a[0],a[1],a[2]))
print('{0} {1} {2}'.format(*a))
```

```python
pt = int( input("점수를 입력하세요") )

if pt >= 900 :
    print(pt, "점은 최상")
    print("축하합니다.")
elif 600 <= pt < 900:
    print(pt, "점은 보통")
else:
    print(pt,"점은 노력 요함")
```
