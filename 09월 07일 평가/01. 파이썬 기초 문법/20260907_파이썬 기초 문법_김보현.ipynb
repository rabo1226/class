2. 매개변수로 두 개의 리스트를 받아, 두 개의 리스트에 저장된 모든 데이터의 합을 리턴하는 함수를 작성하시오.  
EX> 두 매개변수 : [1,2,3] , [4,5,6] / 결과 : 21

```python
#2번 문제 풀이 작성
def sumList(num1, num2) :
  numList = num1 + num2
  numListSum = 0
  for e in numList : 
    numListSum = numListSum + e
  return (f'결과 : {numListSum}')

sumList([1,2,3], [4,5,6])
```

3. 매개변수로 두 개의 리스트를 받아, 두 개의 리스트에 중복으로 저장된 데이터만을 갖는  새로운 리스트를 리턴하는 함수를 작성하시오.  
EX> 두 매개변수 : [1,3,5,7,9] , [1,2,3,4,5] / 결과 : [1,3,5]

```python
#3번 문제 풀이 작성
newList = []

def duList(num1, num2) : 
  for e in num1 : 
    for i in num2 :
      if e == i : 
        newList.append(e)
  return newList

print(f'결과 : {duList([1,3,5,7,9],[1,2,3,4,5])}')
```

6. 아래 딕셔너리 데이터에 stu_grade라는 키를 추가하고, 해당 키의  데이터로  점수(stu_score)에 따른 등급 데이터를 저장하시오.  
90 <= 점수      : A등급  
80 <= 점수 < 90 : B등급  
70 <= 점수 < 80 : C등급  
      점수 < 70 : D등급 

```python
class_room_data = {
    'stu_no' : [1, 2, 3, 4, 5],
    'stu_name' : ['김자바', '이자바', '박자바', '최자바', '정자바'],
    'stu_score' : [85, 65, 70, 80, 95]
}

#6번 문제 풀이 작성
stu_grade = []

for e in class_room_data['stu_score'] : 
    if e >= 90 :
        stu_grade.insert(e, 'A등급')
    elif e <= 89 and e >= 80 :
        stu_grade.insert(e, 'B등급')
    elif e <= 79 and e >= 70 :
        stu_grade.insert(e, 'C등급')
    elif e <= 69 :
        stu_grade.insert(e, 'D등급')

class_room_data['stu_grade'] = stu_grade

print(class_room_data)
```
